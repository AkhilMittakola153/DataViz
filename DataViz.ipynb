```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
tips = sns.load_dataset("tips")
print(tips.head())
print(tips.dtypes)
 
sns.set_theme(style="whitegrid")         
palette = "viridis"                       
```

```python
# 1. HISTOGRAM — distribution of total_bill
# ==============================================================
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
 
# --- Matplotlib ---
axes[0].hist(tips["total_bill"], bins=20, color="#4C72B0", edgecolor="white")
axes[0].set_title("Total Bill Distribution (Matplotlib)", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Total Bill ($)", fontsize=11)
axes[0].set_ylabel("Frequency", fontsize=11)
 
# --- Seaborn ---
sns.histplot(data=tips, x="total_bill", bins=20, kde=True, color="#55A868", ax=axes[1])
axes[1].set_title("Total Bill Distribution (Seaborn)", fontsize=13, fontweight="bold")
axes[1].set_xlabel("Total Bill ($)", fontsize=11)
axes[1].set_ylabel("Frequency", fontsize=11)
 
plt.tight_layout()
plt.savefig("C:/Users/hp/Desktop/Akhil/Project/DataViz/01_histogram.png", dpi=150)
plt.close()
```

```python
# 2. BAR CHART — average total_bill by day
# ==============================================================
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
 
day_order = ["Thur", "Fri", "Sat", "Sun"]
avg_bill_by_day = tips.groupby("day", observed=True)["total_bill"].mean().reindex(day_order)
 
# --- Matplotlib ---
axes[0].bar(avg_bill_by_day.index, avg_bill_by_day.values,
            color=sns.color_palette(palette, len(avg_bill_by_day)))
axes[0].set_title("Average Total Bill by Day (Matplotlib)", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Day", fontsize=11)
axes[0].set_ylabel("Average Total Bill ($)", fontsize=11)
axes[0].yaxis.set_major_formatter(mticker.FormatStrFormatter("$%.0f"))
 
# --- Seaborn ---
sns.barplot(data=tips, x="day", y="total_bill", order=day_order,
            hue="day", palette=palette, ax=axes[1])
axes[1].set_title("Average Total Bill by Day (Seaborn)", fontsize=13, fontweight="bold")
axes[1].set_xlabel("Day", fontsize=11)
axes[1].set_ylabel("Average Total Bill ($)", fontsize=11)
 
plt.tight_layout()
plt.savefig("C:/Users/hp/Desktop/Akhil/Project/DataViz/02_barchart.png", dpi=150)
plt.close()
```

```python
# 3. BOX PLOT — total_bill distribution by day, split by smoker
# ==============================================================
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
 
# --- Matplotlib ---
groups = [tips.loc[tips["day"] == d, "total_bill"].values for d in day_order]
bp = axes[0].boxplot(groups, labels=day_order, patch_artist=True)
colors = sns.color_palette(palette, len(day_order))
for patch, color in zip(bp["boxes"], colors):
    patch.set_facecolor(color)
axes[0].set_title("Total Bill by Day (Matplotlib)", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Day", fontsize=11)
axes[0].set_ylabel("Total Bill ($)", fontsize=11)
 
# --- Seaborn ---
sns.boxplot(data=tips, x="day", y="total_bill", order=day_order,
            hue="smoker", palette="Set2", ax=axes[1])
axes[1].set_title("Total Bill by Day, split by Smoker (Seaborn)", fontsize=13, fontweight="bold")
axes[1].set_xlabel("Day", fontsize=11)
axes[1].set_ylabel("Total Bill ($)", fontsize=11)
axes[1].legend(title="Smoker")
 
plt.tight_layout()
plt.savefig("C:/Users/hp/Desktop/Akhil/Project/DataViz/03_boxplot.png", dpi=150)
plt.close()
```

```python
# 4. SCATTER PLOT — total_bill vs tip
# ==============================================================
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
 
# --- Matplotlib ---
sizes = tips["size"] * 20
scatter = axes[0].scatter(tips["total_bill"], tips["tip"],
                           c=tips["size"], cmap=palette, s=sizes,
                           alpha=0.7, edgecolor="white")
axes[0].set_title("Total Bill vs Tip (Matplotlib)", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Total Bill ($)", fontsize=11)
axes[0].set_ylabel("Tip ($)", fontsize=11)
cbar = plt.colorbar(scatter, ax=axes[0])
cbar.set_label("Party Size")
 
# --- Seaborn ---
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time",
                 style="smoker", size="size", sizes=(20, 200),
                 palette="deep", ax=axes[1])
axes[1].set_title("Total Bill vs Tip (Seaborn)", fontsize=13, fontweight="bold")
axes[1].set_xlabel("Total Bill ($)", fontsize=11)
axes[1].set_ylabel("Tip ($)", fontsize=11)
axes[1].legend(bbox_to_anchor=(1.02, 1), loc="upper left", fontsize=8)
 
plt.tight_layout()
plt.savefig("C:/Users/hp/Desktop/Akhil/Project/DataViz/04_scatterplot.png", dpi=150)
plt.close()
```

```python
# 5. HEATMAP — correlation matrix of numeric columns
# ==============================================================
numeric_cols = tips.select_dtypes(include=np.number)
corr = numeric_cols.corr()
 
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
 
# --- Matplotlib ---
im = axes[0].imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
axes[0].set_xticks(range(len(corr.columns)))
axes[0].set_yticks(range(len(corr.columns)))
axes[0].set_xticklabels(corr.columns, rotation=45, ha="right")
axes[0].set_yticklabels(corr.columns)
for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        axes[0].text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center",
                     color="black", fontsize=9)
axes[0].set_title("Correlation Heatmap (Matplotlib)", fontsize=13, fontweight="bold")
plt.colorbar(im, ax=axes[0])
 
# --- Seaborn ---
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1,
            square=True, ax=axes[1], cbar_kws={"shrink": 0.8})
axes[1].set_title("Correlation Heatmap (Seaborn)", fontsize=13, fontweight="bold")
 
plt.tight_layout()
plt.savefig("C:/Users/hp/Desktop/Akhil/Project/DataViz/05_heatmap.png", dpi=150)
plt.close()
```
